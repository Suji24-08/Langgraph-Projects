Student Report Generator Agent

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict


class Stud_state(TypedDict):
    name : str
    std : str
    marks : list
    total : int
    average : float
    grade : str
    report : str
    
def calc_total(state : Stud_state):
    # tot = 0
    # for mark in state["marks"]:
    #     tot += mark
    # return {state["total"] : tot}
    # return {"total" : tot}
    return {"total" : sum(state["marks"])}

def calc_avg(state : Stud_state):
    # count = 0
    # for mark in state["marks"]:
    #     count +=1
    # return {state["average"] : state["total"]/count}
    # return {"average" : state["total"]/count}
    return {"average" : state["total"]/len(state["marks"])}

def assign_grade(state : Stud_state):
    grade = ""
    if state["average"] >= 90:
        grade = "A"
    elif state["average"] >= 80:
        grade = "B"
    elif state["average"] >= 70:
        grade = "C"
    elif state["average"] >= 60:
        grade = "D"
    else :
        grade = "F"
    # return {state["grade"] : grade}
    return {"grade" : grade}
    
def gen_report(state : Stud_state):
    repo = f"Name   : {state['name'] }\nstd : {state['std']}\nTotal Marks: {state['total']}\nAverage: {state['average']:.2f}\nGrade: {state['grade']}"
    # return {state["report"]: repo}
    return {"report": repo}


builder = StateGraph(Stud_state)

builder.add_node("calc_total",calc_total)
builder.add_node("calc_avg",calc_avg)
builder.add_node("assign_grade",assign_grade)
builder.add_node("gen_report",gen_report)

builder.add_edge(START,"calc_total")
builder.add_edge("calc_total","calc_avg")
builder.add_edge("calc_avg","assign_grade")
builder.add_edge("assign_grade","gen_report")
builder.add_edge("gen_report",END)

graph = builder.compile()


result = graph.invoke({"name":"Sujitha","std":"10th","marks":[88,98,78,86,98]})
print(result["report"])
```
