Password Analyser Agent

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class Pass_State(TypedDict):
    password : str
    length : str
    numbers : str
    spc_char : str
    result : str
    strength : str
    
def check_length(state : Pass_State):
    return {"length": "OK" if len(state["password"])> 5 else "No"}

def check_numbers(state : Pass_State):
    return {"numbers": "Yes" if(any(char.isdigit() for char in state["password"])) else "No"}

def check_spc_char(state : Pass_State):
    return {"spc_char": "Yes" if(any(not char.isalnum() and not char.isspace() for char in state["password"])) else "No"}

def check_strength(state : Pass_State):
    strength = ""
    if(state["length"] =="OK" and state["numbers"]=="Yes" and state["spc_char"]=="Yes"):
        strength = "Strong"
    elif(state["length"] =="OK" and state["numbers"]=="Yes" or state["spc_char"]=="Yes"):
        strength = "Moderate"
    else:
        strength = "Weak"
    return {"strength" : strength}

def result(state : Pass_State):
    return {"result" :  f"==== Password Analyser ====\nStrength = {state['strength']}\nLength   : {state['length']}\nNumber : {state['numbers']}\nSpecial Character : {state['spc_char']}"}

builder = StateGraph(Pass_State)
builder.add_node("check_length",check_length)
builder.add_node("check_numbers",check_numbers)
builder.add_node("check_spc_char",check_spc_char)
builder.add_node("check_strength",check_strength)
builder.add_node("result",result)

builder.add_edge(START,"check_length")
builder.add_edge("check_length","check_numbers")
builder.add_edge("check_numbers","check_spc_char")
builder.add_edge("check_spc_char","check_strength")
builder.add_edge("check_strength","result")
builder.add_edge("result",END)

graph = builder.compile()

res = graph.invoke({"password" : "Sujithanas12!"})
print(res["result"])
```
