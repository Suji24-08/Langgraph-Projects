Email Analyser Agent


```python
import re
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class Email_State(TypedDict):
    email : str
    clean_email : str
    username : str
    domain : str
    report : str
    
def clean(state : Email_State):
    return {"clean_email": (re.sub(r"\s+","",state["email"].strip()).lower())}

def user_name(state : Email_State):
    name = state["clean_email"].split("@")
    return {"username" : name[0]}

def domain(state : Email_State):
    domain = state["clean_email"].split("@")
    return {"domain" : domain[1]}

def report(state : Email_State):
    return {"report" : f"====Email Analyser ===\n Username : {state["username"]}\nDomain : {state["domain"]}"}

builder = StateGraph(Email_State)
builder.add_node("clean",clean)
builder.add_node("username",user_name)
builder.add_node("domain",domain)
builder.add_node("report",report)

builder.add_edge(START,"clean")
builder.add_edge("clean","username")
builder.add_edge("username","domain")
builder.add_edge("domain","report")
builder.add_edge("report",END)

graph = builder.compile()
result = graph.invoke({"email": "   suji 12 98 @ gm ai  l.com     "})
result1 = graph.invoke({"email": "   raj esh wari 89 @ ya hoo  .com     "})
result2 = graph.invoke({"email": "   aru n 90 @ main . c   o.in     "})

print(result["report"])
print(result1["report"])
print(result2["report"])
```
